# 1. 데이터 셋

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/4aix/data/refs/heads/master/ch13_apt_fillna_median.csv'
df = pd.read_csv(url)
df.sampleple()
```

- 지역명2 : 지역명필드를 라벨인코딩하여 추가
- 독립변수(X) : 지역명2, 연도, 월
- 종속변수(y) : 평당분양가격
- 독립변수와 종속변수(2차원으로 reshape)의 스케일 조정

    * 정규화(MinMaxScaler) 작업후 : 지역명2m, 연도m, 월m 컬럼, 평당분양가격m => df_m
    * 표준화(StandardScaler) 작업후 : 지역명2s, 연도s,월s 컬럼, 평당분양가격s => df_s

    => 지역명, 연도, 월, 지역명2, 지역명2m, 연도m, 월m, 평당분양가격m, 지역명2s, 연도s, 월s, 평당분양가격s

y = w1*x1* = w2*x2 +w3*x3 +b

```python
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
```

# 2. 지역명의 라벨 인코딩

```python
le = LabelEncoder()
df['지역명2'] = le.fit_transform(df['지역명'])
df.head(18)
```

# 3. MinMaxScaling

```python
X = df[['지역명2', '연도', '월']]
y = df[['평당분양가격']]
X.shape, y.shape
```

```python
scaler_x = MinMaxScaler()
df[['지역명2m', '연도m', '월m']] = scaler_x.fit_transform(X)

scaler_y = MinMaxScaler()
df['평당분양가격m'] = scaler_y.fit_transform(y)
df.head()
```

# 4. StandatdScaling

```python
scaler_std_x = StandardScaler()
df[['지역명2s', '연도s', '월s']] = scaler_std_x.fit_transform(X)
scaler_std_y = StandardScaler()
y_data = df['평당분양가격'].values.reshape(-1,1)
df['평당분양가격s'] = scaler_std_y.fit_transform(y_data)
df.head()
```

# 5. 지역명을 원핫인코딩

```python
list(df.loc[:16].sort_values(by='지역명2')['지역명'])
```

```python
loc = list(df.loc[:16].sort_values(by='지역명2')['지역명'])
print('원핫인코딩 후 열이름 :', loc)
```

```python
from tensorflow.keras.utils import to_categorical
temp1 = to_categorical(df['지역명2'])
temp2 = pd.get_dummies(df['지역명2']).values
np.all(temp1==temp2)
```

```python
# 원핫인코딩 방법1 : to_categorical
df[loc] = to_categorical(df['지역명2'])
df.head()
```

```python
df.drop(loc, axis=1, inplace=True)
```

```python
# 원핫인코딩 방법2 : get_dummies
df[loc] = pd.get_dummies(df['지역명2'])
df.head()
```
